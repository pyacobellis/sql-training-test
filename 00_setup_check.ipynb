```python
%%sql
SELECT 'facilities' AS table_name, COUNT(*) AS row_count FROM facilities
UNION ALL SELECT 'diagnosis_codes', COUNT(*) FROM diagnosis_codes
UNION ALL SELECT 'patients', COUNT(*) FROM patients
UNION ALL SELECT 'admissions', COUNT(*) FROM admissions;
```

```python
%%sql
SELECT f.facility_name, COUNT(*) AS admissions
FROM admissions a
JOIN facilities f ON f.facility_id = a.facility_id
GROUP BY f.facility_name
ORDER BY admissions DESC;
```
